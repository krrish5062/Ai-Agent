# Creating a Simple Agent with Tracing

```python
import dotenv
import os

from openai import OpenAI

dotenv.load_dotenv()

if not os.environ.get("OPENAI_API_KEY"):
    print(
        """Error: OPENAI_API_KEY environment variable not set. Please copy the .env.template file as .env and fill it in.
    
    You can execute these commands in the terminal to get started:
    cp .env.template .env
    code .env
    """
    )

# Test OpenAI Access
print(
    OpenAI()
    .responses.create(
        model=os.environ["OPENAI_DEFAULT_MODEL"], input="Say: We are up and running!"
    )
    .output_text
)
```

```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

Study Coach agent


```python
study_agent = Agent(
    name="Study Coach",
    instructions="""
    Help students understand topics by explaining them clearly and in small steps.
    Ask what they are learning and their current level. Give examples, then offer
    a short quiz to check understanding. If they get an answer wrong, explain the
    mistake kindly and try a different example.
    """
)
```

Let's execute the Agent:

```python
with trace("Study Coach"):
    result = await Runner.run(
        study_agent,
        "Explain photosynthesis simply, then quiz me with one question."
    )

print(result.final_output)
```

Streaming the answer to the screen, token by token

```python
response_stream = Runner.run_streamed(
    study_agent,
    "Explain photosynthesis simply, then quiz me with one question."
)

async for event in response_stream.stream_events():
    if event.type == "raw_response_event" and isinstance(
        event.data, ResponseTextDeltaEvent
    ):
        print(event.data.delta, end="", flush=True)
```
